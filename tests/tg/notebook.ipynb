```python
from telegram.ext import Application
from telegram import InlineKeyboardMarkup, InlineKeyboardButton
import asyncio

application = (
    Application.builder()
    .token(YOUR_TOKEN)
    .build()
)

import app.data.emoji as emoji

def get_test_keyboard():
    return InlineKeyboardMarkup([
         [
             InlineKeyboardButton(f"{emoji.ANCHOR} Create route", callback_data="1"),
             InlineKeyboardButton(f"{emoji.STATS_LINE} Port prices", callback_data="1"),
             InlineKeyboardButton(f"{emoji.NOTE} Create supplier request", callback_data="1"),
         ],
        [
             InlineKeyboardButton(f"{emoji.MAP} Show my routes ", callback_data="1"),
             InlineKeyboardButton(f" ", callback_data="1"),
             InlineKeyboardButton(f"{emoji.DOC} Show requests", callback_data="1"),
        ]
    ])

async def run():
    print("🔧 Starting bot in async polling mode...")

    await application.initialize()
    await application.start()

    await application.bot.send_message(
        chat_id=501327560,
        text="Main menu navigation",
        reply_markup=get_test_keyboard()
    )

    # keep bot alive
    await application.updater.start_polling()

await run()
```
